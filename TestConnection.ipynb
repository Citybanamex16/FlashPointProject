```python
import threading
from flask import Flask, jsonify, request
from flask_cors import CORS

# 1. Creación de la app web
app = Flask(__name__)
CORS(app) # Habilitar llamadas desde cualquier origen (incluyendo Unity)

# 2. Definición de la Ruta / Endpoint
@app.route('/api/process', methods=['POST'])

# 2.1 Esta es el middleware que manda a ejecutar el enrutador cuando recibe una petición.
def process_data():
    # A. Recepción de datos desde Unity
    data = request.get_json()
    
    # Extraemos información que envía Unity (con un valor por defecto si falla)
    player_name = data.get('playerName', 'Jugador Anónimo')
    score = data.get('score', 0)
    
    print(f"[PYTHON]: Datos recibidos de Unity -> Jugador: {player_name}, Score: {score}")
    
    # B. Lógica de negocio / IA / Algoritmo pesado
    calculated_level = score + 1
    custom_message = f"¡Hola {player_name}! Tu nivel calculado en Python es {calculated_level}."
    
    # C. Estructuración del JSON de respuesta (El Modelo de Python)
    response_payload = {
        "status": "success",
        "playerName": player_name,
        "calculatedLevel": calculated_level,
        "message": custom_message
    }
    
    # D. Envío en formato JSON con código de estado 200 (OK)
    return jsonify(response_payload), 200


# 3. Estrategia de Ejecución en Hilos (Thread) para Jupyter Notebook
def run_server():
    # Runs backend on port 5000
    app.run(host='127.0.0.1', port=5000, debug=False, use_reloader=False)

# Levantamos el servidor en un hilo secundario
server_thread = threading.Thread(target=run_server)
server_thread.daemon = True
server_thread.start()

print("🚀 ¡Servidor escuchando en http://127.0.0.1:5000/api/process!")
```
